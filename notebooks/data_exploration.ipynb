```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("../data/raw/heart.csv")
```

```python
print(df.shape)


print(df.head())


print(df.tail())


print(df.columns)


print(df.dtypes)


print(df.isnull().sum())


print(df.describe())

# Check target column distribution
print(df['target'].value_counts())
```

```python
# Chart 1 - How many people have heart disease vs not
plt.figure(figsize=(6, 4))
sns.countplot(x='target', data=df)
plt.title("Heart Disease Count")
plt.xticks([0, 1], ['No Disease', 'Disease'])
plt.savefig("../reports/figures/01_target_distribution.png")
plt.show()
```

```python

# Chart 2 - Age distribution
plt.figure(figsize=(6, 4))
plt.hist(df['age'], bins=20, color='steelblue', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.savefig("../reports/figures/02_age_distribution.png")
plt.show()
```

```python
# Chart 3 - Age vs Heart Disease
plt.figure(figsize=(6, 4))
sns.boxplot(x='target', y='age', data=df)
plt.title("Age vs Heart Disease")
plt.xticks([0, 1], ['No Disease', 'Disease'])
plt.savefig("../reports/figures/03_age_vs_target.png")
plt.show()
```

```python
# Chart 4 - Cholesterol vs Heart Disease
plt.figure(figsize=(6, 4))
sns.boxplot(x='target', y='chol', data=df)
plt.title("Cholesterol vs Heart Disease")
plt.xticks([0, 1], ['No Disease', 'Disease'])
plt.savefig("../reports/figures/04_chol_vs_target.png")
plt.show()
```

```python
# Chart 5 - Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.savefig("../reports/figures/05_correlation_heatmap.png")
plt.show()
```

```python
# Chart 6 - Max heart rate vs Heart Disease
plt.figure(figsize=(6, 4))
sns.boxplot(x='target', y='thalach', data=df)
plt.title("Max Heart Rate vs Heart Disease")
plt.xticks([0, 1], ['No Disease', 'Disease'])
plt.savefig("../reports/figures/06_thalach_vs_target.png")
plt.show()
```
